```python
from pathlib import Path
import re
import json

files = Path('../catalogue/400-filenames.txt').read_text()

files = [{'path': f.strip()} for f in files.split('\n') if f.strip()]

print(json.dumps(files[:3], indent=2))
```

```python
for f in files:
    f['filename'] = re.sub(r'^.*\/', '', f['path'])
    f['folder'] = f['path'].split('/')[1]
    f['code'] = re.findall(r'\d+', f['filename'])[0]
    f['filter'] = re.findall(r'^.*?\d+\D', f['filename'])[0]

print(json.dumps(files[:3], indent=2))
```

```python
# Recursive random sampling from the file list: n=20, 15 and 10
# output: filters: a list of filters we can plug into FrameSense collection.json

import random
random.seed(1234)

sizes = [20, 15, 10]
samples = {}
sample = files
filters = {}
for s in sizes:
    sample = random.sample(sample, s)
    samples[s] = sample
    filters[f'sample{s}'] = '|'.join([f['filter'] for f in sample])

print(json.dumps(filters, indent=2))
```
